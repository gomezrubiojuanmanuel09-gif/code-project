```r
# ===================================================================
# PROYECTO: VISUALIZACIÓN DE COORDENADAS Y DISTANCIAS GENÉTICAS
# OBJETIVO: Mapear poblaciones (Azteca, Irlandés) en un plano cartesiano (2D)
# --- PASO 1: DEFINICIÓN DE DATOS ---
# Creamos una matriz con coordenadas simuladas para simular un análisis
# de componentes principales (PCA). El origen (0,0) representa al ancestro común P.
nombres_poblaciones <- c("Ancestro_P", "Azteca_Q", "Irlandes_R", "Celta_Ref", "Germanico_Ref")

coordenadas_x <- c(0.00,  0.42, -0.15, -0.12, -0.08) # Componente 1 (Eje X)
coordenadas_y <- c(0.00, -0.18,  0.35,  0.31,  0.27) # Componente 2 (Eje Y)

# --- PASO 2: CONFIGURACIÓN DEL ESPACIO DE DIBUJO ---
# Ajustamos los márgenes del gráfico para que los textos no se corten
par(mar = c(5, 5, 4, 2))

# Creamos un lienzo en blanco con las dimensiones correctas
plot(coordenadas_x, coordenadas_y,
     type = "n",                             # "n" significa invisible (no dibuja los puntos aún)
     xlim = c(-0.4, 0.6),                    # Límites del eje X
     ylim = c(-0.4, 0.6),                    # Límites del eje Y
     xlab = "Componente Genético 1 (Deriva)",
     ylab = "Componente Genético 2 (Estructura)",
     main = "Mapa de Distancias Filogenéticas Coordenadas",
     font.main = 2,                          # Título en negrita
     font.lab = 2,                           # Etiquetas de ejes en negrita
     las = 1)                                # Números del eje Y en posición horizontal

# --- PASO 3: ELEMENTOS VISUALES DE REFERENCIA ---
# Dibujamos líneas de cuadrícula suaves de fondo para facilitar la lectura
grid(nx = NULL, ny = NULL, col = "gray90", lty = "dotted", lwd = 1)

# Dibujamos los ejes centrales (X=0 e Y=0) para marcar el punto de partida evolutivo
abline(h = 0, v = 0, col = "gray70", lty = "dashed")

# --- PASO 4: DIBUJO DE LÍNEAS DE CONEXIÓN EVOLUTIVA ---
# Trazamos líneas que van desde el ancestro común (P) hacia las poblaciones derivadas
# Esto ayuda a visualizar la distancia lineal recta.
arrows(0, 0, 0.42, -0.18, col = "#b91c1c", lwd = 2, length = 0.1) # Flecha hacia Azteca (Rojo)
arrows(0, 0, -0.15, 0.35, col = "#1d4ed8", lwd = 2, length = 0.1) # Flecha hacia Irlandés (Azul)

# --- PASO 5: DIBUJO DE LOS PUNTOS COORDENADOS ---
# Asignamos colores específicos a cada población para identificarlas visualmente
colores_puntos <- c("#475569", "#b91c1c", "#1d4ed8", "#059669", "#d97706")

# Dibujamos los puntos en el mapa
points(coordenadas_x, coordenadas_y,
       pch = 19,         # Código 19 representa círculos sólidos rellenos
       cex = 2,          # Tamaño de los círculos
       col = colores_puntos)

# --- PASO 6: AGREGAR ETIQUETAS DE TEXTO ---
# Colocamos los nombres de cada población al lado de sus respectivos puntos
# Usamos un pequeño desplazamiento (pos = 3 significa arriba, pos = 4 a la derecha)
text(coordenadas_x[1], coordenadas_y[1], labels = nombres_poblaciones[1], pos = 3, font = 2, cex = 0.9)
text(coordenadas_x[2], coordenadas_y[2], labels = nombres_poblaciones[2], pos = 4, font = 2, cex = 0.9, col = "#b91c1c")
text(coordenadas_x[3], coordenadas_y[3], labels = nombres_poblaciones[3], pos = 2, font = 2, cex = 0.9, col = "#1d4ed8")
text(coordenadas_x[4], coordenadas_y[4], labels = nombres_poblaciones[4], pos = 4, cex = 0.8)
text(coordenadas_x[5], coordenadas_y[5], labels = nombres_poblaciones[5], pos = 3, cex = 0.8)

# --- PASO 7: AGREGAR NOTAS ACLARATORIAS ---
# Añadimos una pequeña leyenda en la esquina inferior izquierda para dar contexto formal
legend("bottomleft",
       legend = c("Origen (Haplogrupo P)", "Rama Americana (Q)", "Rama Europea (R)"),
       col = c("#475569", "#b91c1c", "#1d4ed8"),
       pch = 19,
       bty = "o",        # "o" dibuja una caja alrededor de la leyenda
       bg = "white",     # Fondo blanco para que no se mezcle con la cuadrícula
       box.col = "gray80",
       cex = 0.8)
```

```r
# ===================================================================
# MAPA DE DIVERGENCIA DIOFÁNTICA: HAPLOGRUPOS Q y R
# Basado en el Teorema de Euclides y la Identidad de Bézout
# ===================================================================

# --- 1. PARÁMETROS DEL TEOREMA ---
# Ecuación: 13x + 8y = 1
a <- 13   # Tasa de mutación Linaje R
b <- 8    # Tasa de mutación Linaje Q
x0 <- -3  # Solución particular (Eje X) calculada con Euclides
y0 <- 5   # Solución particular (Eje Y) calculada con Euclides

# Generamos un rango de soluciones generales usando la variable 'k'
k_valores <- c(-2, -1, 0, 1, 2)
nodos_x <- x0 + b * k_valores
nodos_y <- y0 - a * k_valores

# --- 2. CONFIGURACIÓN DEL LIENZO ---
par(mar = c(5, 5, 4, 2))

# Creamos el espacio de dibujo ajustado a nuestras coordenadas matemáticas
plot(nodos_x, nodos_y,
     type = "n",
     xlim = c(-25, 20),
     ylim = c(-25, 35),
     xlab = "Deriva Generacional Linaje R (x)",
     ylab = "Deriva Generacional Linaje Q (y)",
     main = expression(bold("Mapa de Divergencia Diofántica: 13x + 8y = 1")),
     font.lab = 2,
     las = 1)

# --- 3. REFERENCIAS Y LÍNEAS GUÍA ---
# Cuadrícula suave de fondo
grid(nx = NULL, ny = NULL, col = "gray90", lty = "dotted", lwd = 1)

# Ejes centrales (El origen evolutivo)
abline(h = 0, v = 0, col = "gray60", lty = "dashed")

# Dibujamos la recta matemática continua de la ecuación (y = (1 - 13x) / 8)
curve((1 - 13 * x) / 8, from = -30, to = 30, add = TRUE, col = "#94a3b8", lwd = 1.5)

# --- 4. VECTORES EVOLUTIVOS (FLECHAS) ---
# Flecha desde el Ancestro P (0,0) hasta el estado óptimo k=0 (-3, 5)
arrows(0, 0, x0, y0, col = "#b91c1c", lwd = 2.5, length = 0.12)

# --- 5. DIBUJO DE LOS PUNTOS ---
# 5.1 Dibujamos todas las soluciones enteras posibles (Nodos grises)
points(nodos_x, nodos_y, pch = 19, cex = 1.5, col = "#64748b")

# 5.2 Marcamos el Ancestro Común en el origen
points(0, 0, pch = 15, cex = 1.5, col = "#1e293b") # Cuadrado oscuro

# 5.3 Resaltamos el estado óptimo (Solución k=0)
points(x0, y0, pch = 19, cex = 2, col = "#b91c1c") # Círculo rojo grande

# --- 6. ETIQUETAS DE TEXTO HUMANIZADAS ---
# Etiqueta del Ancestro
text(0, 0, labels = "Ancestro P (0,0)", pos = 4, font = 2, cex = 0.85, col = "#1e293b")

# Etiquetas de los nodos diofánticos
for(i in 1:length(k_valores)) {
  # Solo etiquetamos los nodos que no son el principal para no amontonar
  if(k_valores[i] != 0) {
    text(nodos_x[i], nodos_y[i],
         labels = paste0("k=", k_valores[i], "\n(", nodos_x[i], ",", nodos_y[i], ")"),
         pos = 2, cex = 0.7, col = "#475569")
  }
}

# Etiqueta del estado evolutivo óptimo (k=0)
text(x0, y0, labels = "Estado Óptimo k=0\n(-3, 5)", pos = 3, font = 2, cex = 0.9, col = "#b91c1c")

# --- 7. LEYENDA ---
legend("topright",
       legend = c("Recta 13x + 8y = 1", "Soluciones Enteras Válidas", "Vector de Deriva Óptimo", "Origen Ancestral"),
       col = c("#94a3b8", "#64748b", "#b91c1c", "#1e293b"),
       lty = c(1, NA, 1, NA),
       pch = c(NA, 19, NA, 15),
       lwd = c(1.5, NA, 2.5, NA),
       bty = "o", bg = "white", box.col = "gray80", cex = 0.8)
```

```r
# Funcion recursiva para ejecutar el Algoritmo de Euclides Extendido
euclides_extendido <- function(a, b) {
if (b == 0) {
return(list(mcd = a, x = 1, y = 0))
} else {
resultado <- euclides_extendido(b, a %% b)
x_actual <- resultado$y
y_actual <- resultado$x - (a %/% b) * resultado$y
return(list(mcd = resultado$mcd, x = x_actual, y = y_actual))
}
}
# Solucionador formal de Ecuaciones Diofánticas Lineales Genéticas
resolver_diofantica_genetica <- function(a, b, c) {
res_euclides <- euclides_extendido(a, b)
mcd_calc <- res_euclides$mcd
# Verificacion de la condicion del Teorema de Existencia
if (c %% mcd_calc != 0) {
stop(paste("Error: La distancia genómica", c, "no es almacenable discretamente.",
"MCD(", a, ",", b, ") =", mcd_calc, "no divide a", c))
}
# Escalado de los coeficientes de Bezout para la distancia objetivo c
factor_escala <- c / mcd_calc
x_particular <- res_euclides$x * factor_escala
y_particular <- res_euclides$y * factor_escala
# Construccion de los parametros de la solucion general
cat("=========================================================\n")
cat(" SOLUCIONADOR DIOFÁNTICO DE LINAJES \n")
cat("=========================================================\n")
cat("Ecuación planteada:", a, "x +", b, "y =", c, "\n")
cat("Máximo Común Divisor mcd(", a, ",", b, ") =", mcd_calc, "\n")
cat("Solución Particular encontrada: x0 =", x_particular, ", y0 =", y_particular, "\n\n")
cat("Fórmulas de Estabilidad General (Solución General):\n")
cat("x =", x_particular, "+", (b / mcd_calc), "k\n")
cat("y =", y_particular, "-", (a / mcd_calc), "k\n")
cat("=========================================================\n")
return(list(x0 = x_particular, y0 = y_particular, mcd = mcd_calc))
}
marcador_mutacional <- resolver_diofantica_genetica(a = 13, b = 8, c = 1)
```
